## 3. Prima técnica: aplicar los recargos

La prima técnica carga sobre la pura los recargos que **sí se justifican**: gastos, comisiones,
dividendos, costo de IBNR, costo de reaseguro y margen técnico. Cuando se expresan como **% de la prima**,
la fórmula es:

$$\mathrm{Prima\ técnica}=\frac{\mathrm{Prima\ pura}+\text{cargos fijos}}{1-(g+m)}$$

donde $g$ = gastos+comisiones+dividendos+IBNR+reaseguro (como % de prima) y $m$ = margen.

```python
# ─── Recargos de la prima técnica (como % de la prima) ─────────────────────
recargos = {
    "gastos_operacion": 0.12,   # 12%
    "comisiones":       0.10,   # 10%
    "dividendos":       0.02,   # 2%
    "costo_ibnr":       0.03,   # 3%  (del track de reservas)
    "costo_reaseguro":  0.04,   # 4%  (del track de reaseguro)
}
margen = 0.08                    # 8% margen técnico (seguridad + utilidad)

g = sum(recargos.values())
prima_tecnica = prima_pura / (1 - (g + margen))

print(f"Suma de recargos g = {g:.0%} | margen m = {margen:.0%}")
print(f"Prima técnica = {prima_pura:,.2f} / (1 - {g+margen:.2f}) = ${prima_tecnica:,.2f}")
print(f"→ La prima técnica es {prima_tecnica/prima_pura:.2f}× la prima pura.")
```

## 4. Prima comercial: el criterio de mercado

Sobre la técnica, el área comercial aplica su criterio: un **descuento de mercado** (por competencia o
elasticidad) y, si el cliente paga a plazos, un **recargo por pago fraccionado**. Esto **ya no es
técnico**: es decisión de negocio.

```python
# ─── De técnica a comercial ────────────────────────────────────────────────
descuento_mercado = 0.10        # 10% de descuento por competencia (decisión comercial)
prima_comercial = prima_tecnica * (1 - descuento_mercado)

# recargo por pago fraccionado (si paga en n exhibiciones)
n_pagos = 12
recargo_frac = 0.06             # 6% por fraccionar
prima_comercial_frac = prima_comercial * (1 + recargo_frac)
pago_mensual = prima_comercial_frac / n_pagos

print(f"Prima técnica            : ${prima_tecnica:,.2f}")
print(f"− descuento de mercado 10%: ${prima_comercial:,.2f}   (prima comercial de contado)")
print(f"+ recargo fraccionado 6% : ${prima_comercial_frac:,.2f}   (si paga en {n_pagos} exhibiciones)")
print(f"  → pago mensual          : ${pago_mensual:,.2f}")
print(f"\nOjo: la comercial ({prima_comercial:,.0f}) quedó por DEBAJO de la técnica "
      f"({prima_tecnica:,.0f}): es un riesgo de negocio consciente.")
```

## 5. Armar el recibo

El recibo es el desglose que ve el cliente. Sobre la prima comercial se suman los **derechos de póliza**
y el **IVA (16%)** para llegar a la **prima total**.

```python
# ─── El recibo completo ────────────────────────────────────────────────────
derechos_poliza = 150.0
iva_tasa = 0.16

base_iva = prima_comercial_frac + derechos_poliza
iva = base_iva * iva_tasa
prima_total = base_iva + iva

print("─" * 42)
print("  RECIBO — desglose de la prima")
print("─" * 42)
print(f"  Prima comercial (fracc.) : ${prima_comercial_frac:>10,.2f}")
print(f"  Derechos de póliza       : ${derechos_poliza:>10,.2f}")
print(f"  Subtotal                 : ${base_iva:>10,.2f}")
print(f"  IVA (16%)                : ${iva:>10,.2f}")
print("─" * 42)
print(f"  PRIMA TOTAL DEL RECIBO   : ${prima_total:>10,.2f}")
print("─" * 42)
```

## 6. Documentar: la tabla-recibo

Cada capa de la prima es un supuesto. La documentamos en una tabla que separa lo **técnico** (va en la
nota técnica, con su fuente) de lo **comercial** (decisión de negocio).

```python
# ─── Tabla-recibo documentada ──────────────────────────────────────────────
tabla = pd.DataFrame([
    ["Prima pura",           prima_pura,             "técnico",  "modelo de severidad y frecuencia (Sesión 9, ajustada por inflación)"],
    ["Gastos de operación",  0.12,                   "técnico",  "estudio de gastos de la compañía"],
    ["Comisiones",           0.10,                   "técnico",  "esquema de comisiones del canal"],
    ["Dividendos",           0.02,                   "técnico",  "condiciones del producto"],
    ["Costo de IBNR",        0.03,                   "técnico",  "método de reservas (Tema 3)"],
    ["Costo de reaseguro",   0.04,                   "técnico",  "contrato de reaseguro (Tema 5)"],
    ["Margen técnico",       0.08,                   "técnico",  "política de suficiencia y utilidad"],
    ["Descuento de mercado", -0.10,                  "comercial","política comercial / competencia"],
    ["Recargo fraccionado",  0.06,                   "comercial","condiciones de pago"],
    ["Derechos de póliza",   derechos_poliza,        "fiscal",   "tarifa de derechos"],
    ["IVA",                  0.16,                    "fiscal",  "Ley del IVA (16%)"],
], columns=["concepto", "valor", "tipo", "fuente / justificación"])

pd.set_option("display.max_colwidth", 60)
print(tabla.to_string(index=False))
print("\nRegla: lo 'técnico' se documenta y valida en la nota técnica ante la CNSF;")
print("lo 'comercial' es decisión de negocio y no requiere sustento actuarial.")
```
