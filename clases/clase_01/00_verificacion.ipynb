## Una tabla y una gráfica de prueba

Confirmamos que podemos manejar datos (pandas) y graficar (matplotlib).

```python
# ─── Mini-cartera de prueba ─────────────────────────────────────────────────
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)
cartera = pd.DataFrame({
    "id_poliza": [f"AUTO-{i:03d}" for i in range(1, 9)],
    "prima":     np.random.randint(3_000, 12_000, 8),
    "siniestros": np.random.poisson(0.4, 8),
})
display(cartera)

plt.figure(figsize=(7, 3.5))
plt.hist(np.random.lognormal(mean=6, sigma=0.9, size=1000), bins=40,
         color="#17A69B", edgecolor="white")
plt.title("Gráfica de prueba — severidad simulada")
plt.xlabel("monto"); plt.ylabel("frecuencia")
plt.show()

print("Si ves la tabla y el histograma, tu entorno funciona. ✓")
```
